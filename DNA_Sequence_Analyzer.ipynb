# DNA Sequence Analyzer

## Project Objective
This project analyzes a DNA sequence entered by the user. It calculates useful information such as nucleotide counts, sequence length, and GC content, AT Content, Reverse complement, Transcription to RNA, Codons, Amino-acid Translation.

## Version 1 Features

- Take DNA input
- Convert to uppercase
- Validate DNA
- Count A, T, G, C
- Calculate GC Content
- Calculate AT Content
- Reverse complement
- Transcription to RNA
- Codons
- Amino-acid Translation
- Display results


```python
#Step 1 : input of the DNA Sequencedna_sequence = get_sequence()
def get_sequence():
  print("===================================")
  print("      DNA Sequence Analyzer")
  print("===================================\n")
  while True:
        dnasequence = input("Enter a DNA Sequence: ").strip()
        if dnasequence == "":
            print("DNA Sequence cannot be empty.")
            print("Please try again.\n")
        else:
            print(f"DNA Sequence Entered = {dnasequence}")
            print()

#Step 2 : Converting the DNA Sequence into upper case
            dnasequence=dnasequence.upper()
            print(f"DNA Sequence Entered = {dnasequence}")
            print()
            return dnasequence

#Step 3 : Checking the valid DNA Sequence
def validate_sequence(dnasequence):
  for base in dnasequence:
    if base not in "ATGC":
      print("Invalid DNA Sequence")
      valid_sequence=False
      break
  else:
    print("Valid DNA Sequence")
    valid_sequence=True
  print()
  return valid_sequence

#Step 4 : Counting the numbers of base pairs
def count_bases(dnasequence,valid_sequence):
  if valid_sequence:
    acount=dnasequence.count("A")
    tcount=dnasequence.count("T")
    gcount=dnasequence.count("G")
    ccount=dnasequence.count("C")

    print(f"A : {acount}")
    print(f"T : {tcount}")
    print(f"G : {gcount}")
    print(f"C : {ccount}")
    print()

  #Step 5: Find GC% Content
def gc_content(dnasequence,valid_sequence):
  if valid_sequence:
    gcount=dnasequence.count("G")
    ccount=dnasequence.count("C")
    total_content=len(dnasequence)
    gc_content=gcount+ccount
    gc_percent=(gc_content/total_content)*100
    print(f"GC Content = {gc_percent:.2f}%")

#Step 6:Find AT% Content
def at_content(dnasequence,valid_sequence):
  if valid_sequence:
    acount=dnasequence.count("A")
    tcount=dnasequence.count("T")
    total_content=len(dnasequence)
    at_content=acount+tcount
    at_percent=(at_content/total_content)*100
    print(f"AT Content = {at_percent:.2f}%")

#Step 7: Reverse Complement
def reverse_complement(dnasequence,valid_sequence):
  if valid_sequence:
    complement = {'A': 'T', 'T': 'A', 'G': 'C', 'C': 'G'}
    reverse_complement = ''.join([complement[base] for base in dnasequence[::-1]])
    print(f"Reverse Complement = {reverse_complement}")
    return reverse_complement

#Step 8: Transcription to RNA
def transcription(dnasequence,valid_sequence):
  if valid_sequence:
    rna_sequence = dnasequence.replace('T', 'U')
    print(f"Transcription to RNA = {rna_sequence}")
    return rna_sequence

#Step 9: Codons
def codons(dnasequence,valid_sequence):
  if valid_sequence:
    print(f"Codons = {[dnasequence[i:i+3] for i in range(0, len(dnasequence), 3)]}")

#Step 10: Amino-acid Translation
def amino_acid_translation(dnasequence,valid_sequence):
  if valid_sequence:
    codon_table = { 'ATA':'I', 'ATC':'I', 'ATT':'I', 'ATG':'M',
    'ACA':'T', 'ACC':'T', 'ACG':'T', 'ACT':'T',
    'AAC':'N', 'AAT':'N', 'AAA':'K', 'AAG':'K',
    'AGC':'S', 'AGT':'S', 'AGA':'R', 'AGG':'R',
    'CTA':'L', 'CTC':'L', 'CTG':'L', 'CTT':'L',
    'CCA':'P', 'CCC':'P', 'CCG':'P', 'CCT':'P',
    'CAC':'H', 'CAT':'H', 'CAA':'Q', 'CAG':'Q',
    'CGA':'R', 'CGC':'R', 'CGG':'R', 'CGT':'R',
    'GTA':'V', 'GTC':'V', 'GTG':'V', 'GTT':'V',
    'GCA':'A', 'GCC':'A', 'GCG':'A', 'GCT':'A',
    'GAC':'D', 'GAT':'D', 'GAA':'E', 'GAG':'E',
    'GGA':'G', 'GGC':'G', 'GGG':'G', 'GGT':'G',
    'TCA':'S', 'TCC':'S', 'TCG':'S', 'TCT':'S',
    'TTC':'F', 'TTT':'F', 'TTA':'L', 'TTG':'L',
    'TAC':'Y', 'TAT':'Y', 'TAA':'_', 'TAG':'_',
    'TGC':'C', 'TGT':'C', 'TGA':'_', 'TGG':'W'}
    amino_acids = [codon_table[dnasequence[i:i+3]] for i in range(0, len(dnasequence), 3)]
    print(f"Amino-acid Translation = {''.join(amino_acids)}")
    return ''.join(amino_acids)




dnasequence = get_sequence()
valid_sequence = validate_sequence(dnasequence)
count_bases(dnasequence,valid_sequence)
gc_content(dnasequence,valid_sequence)
at_content(dnasequence,valid_sequence)
reverse_complement(dnasequence,valid_sequence)
transcription(dnasequence,valid_sequence)
codons(dnasequence,valid_sequence)
amino_acid_translation(dnasequence,valid_sequence)
```
